# Linear_search

```python
def linearsearch(arr,n,key):
  for i in range(n):
    if arr[i]==key:
      return i
  return -1
arr=[20,21,19,18,17]
key=19
n=len(arr)
print("linear search:",linearsearch(arr,n,key))
```

# Binary_search

```python
def binary_search(arr, key):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == key:
            return mid
        elif arr[mid] < key:
            low = mid + 1
        else:
            high = mid - 1

    return -1

arr = [10, 20, 30, 40, 50]
key = 40
result = binary_search(arr, key)
print("Element found at index", result)
```
